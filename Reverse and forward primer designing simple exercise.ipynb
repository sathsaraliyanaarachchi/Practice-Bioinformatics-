### Reverse and forward primer designing simple exercise

```python
dna="ATGCGATCGATCGATCGATCGATCGATCGAATCGATCGATCGATCGATCGATCGAATCGATCGATCG"
forward_primer=dna[:21]
print("Forward Primer:",forward_primer)
```

```python
dna="ATGCGATCGATCGATCGATCGATCGATCGAATCGATCGATCGATCGATCGATCGAATCGATCGATCG"
last_20=dna[-20:]
reversed_20=last_20[::-1]
complement_map = {"A": "T", "T": "A", "G": "C", "C": "G"}
reverse_primer="".join(complement_map[base] for base in reversed_20)
print("Original Last 20:",last_20)
print("Reverse Primer:",reverse_primer)
```
